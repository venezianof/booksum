### Agente Ricerca Medica Profonda
Configurazione dell'agente per l'analisi di bibliografia Medline, abstract e linee guida terapeutiche.

```python
def get_medical_research(topic):
    system_prompt = """Sei un assistente di ricerca medica avanzata.
    Il tuo compito è fornire:
    1. Bibliografia dettagliata (stile Medline/PubMed).
    2. Abstract riassuntivi delle ultime terapie.
    3. Schemi terapeutici e linee guida ufficiali.
    4. Descrizione di immagini/schemi diagnostici.
    Usa un linguaggio tecnico e preciso."""

    try:
        response = client.chat.completions.create(
            model="deepseek-ai/DeepSeek-V4-Pro",
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": f"Esegui una ricerca profonda su: {topic}"}
            ],
            temperature=0.3
        )
        return response.choices[0].message.content
    except Exception as e:
        return f"Errore durante la ricerca: {str(e)}"
```

```python
# Eseguiamo una ricerca approfondita su un tema oncologico moderno
tema_ricerca = "Immunoterapia e inibitori dei checkpoint nel tumore del polmone non a piccole cellule (NSCLC)"
risultato = get_medical_research(tema_ricerca)

print(f"### Risultati Ricerca Medica: {tema_ricerca} ###\n")
print(risultato)
```

### Riassunto: Procedura di Incisione e Drenaggio (Furuncolosi CAE)

1. **Preparazione**: Paziente supino, testa ruotata e illuminazione ottimale (otoscopio o microscopio).
2. **Anestesia**: Infiltrazione locale con lidocaina 1% (senza adrenalina) usando un ago sottile (27 G).
3. **Incisione**: Taglio di 2-3 mm con lama appuntita sulla sommità del punto di fluttuazione, parallelamente all'asse del condotto.
4. **Evacuazione**: Pressione delicata per rimuovere il materiale purulento e lavaggio con soluzione fisiologica.
5. **Zaffaggio**: Inserimento di una striscia di garza iodoformica o tampone antibiotico nel sito chirurgico.
6. **Follow-up**: Rimozione del drenaggio dopo 24-48 ore e inizio di terapia topica antibiotico-steroidea.
