```python
# Use a pipeline as a high-level helper
from transformers import pipeline

pipe = pipeline("image-text-to-text", model="LiquidAI/d1-omni-600M", trust_remote_code=True)
```

```python
# pip install -U transformers accelerate
# Load model directly
from transformers import AutoModel
model = AutoModel.from_pretrained("LiquidAI/d1-omni-600M", trust_remote_code=True, device_map="auto")
```

### Caricamento dell'immagine ed esecuzione dell'inferenza con LiquidAI

```python
from google.colab import files
from PIL import Image
import io

# Carica l'immagine dal computer locale
uploaded = files.upload()

if uploaded:
    # Ottieni il nome del file caricato
    filename = list(uploaded.keys())[0]
    # Apri l'immagine
    image = Image.open(io.BytesIO(uploaded[filename])).convert("RGB")
    display(image)

    # Esegui il modello d1-omni-600M usando la pipeline 'pipe'
    prompt = "Describe this image in detail."
    # La struttura del prompt per i modelli di questo tipo
    output = pipe(image=image, text=prompt)
    print("\nRisultato del modello:")
    print(output)
else:
    print("Nessun file caricato.")
```
