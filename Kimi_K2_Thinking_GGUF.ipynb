## Local Inference on GPU
Model page: https://huggingface.co/unsloth/Kimi-K2-Thinking-GGUF

⚠️ If the generated code snippets do not work, please open an issue on either the [model repo](https://huggingface.co/unsloth/Kimi-K2-Thinking-GGUF)
			and/or on [huggingface.js](https://github.com/huggingface/huggingface.js/blob/main/packages/tasks/src/model-libraries-snippets.ts) 🙏

```python
from llama_cpp import Llama

# Caricamento del modello Kimi-K2.
# Assicurati di eseguire questa cella e attendere il completamento prima di procedere.
llm = Llama.from_pretrained(
    repo_id="unsloth/Kimi-K2-Thinking-GGUF",
    filename="*Q4_K_M.gguf", # Utilizzo di un formato quantizzato per maggiore efficienza in Colab
    n_ctx=2048,
    n_gpu_layers=-1 # Utilizza la GPU se disponibile
)
```

```python
llm.create_chat_completion(
	messages = [
        {"role": "system", "content": "Sei un assistente medico utile e preciso. Rispondi alla domanda fornendo il trattamento più appropriato e una breve spiegazione in italiano."},
        {"role": "user", "content": "Un uomo di 60 anni con uso recente di cocaina si presenta con dolore toracico acuto e difficoltà respiratorie. I suoi parametri vitali mostrano una pressione arteriosa di 195/110 mmHg. Qual è il trattamento immediato più appropriato? Diuretici ACE-inibitori Beta-bloccanti Nitroglicerina Calcio-antagonisti"}
    ],
    temperature=0.2, # Lower values (e.g., 0.1-0.3) make the output more deterministic and focused.
    max_tokens=200,  # Limits the length of the generated response.
    top_p=0.9         # Samples from the smallest possible set of tokens whose cumulative probability exceeds top_p.
)
```
