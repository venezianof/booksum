# Sviluppo Agente di Ricerca Medica Avanzata
Questo modulo implementa un framework per la ricerca clinica, l'analisi di linee guida e la generazione di bibliografia automatizzata.

````python
import datetime

class MedicalResearchAgent:
    def __init__(self, model_engine=None):
        self.model = model_engine
        self.history = []

    def search_medline(self, query):
        """Simulazione di ricerca su Medline/PubMed API"""
        print(f"[Search] Interrogazione database MEDLINE per: {query}...")
        # Qui verrebbe integrata la logica per Bio.Entrez o simili
        return f"Risultati principali per {query} (Meta-analisi e RCT 2020-2024)"

    def generate_clinical_summary(self, topic):
        """Genera uno schema riassuntivo e linee guida"""
        summary_template = f"""
        ### Schema Riassuntivo: {topic}
        1. **Diagnostica**: Criteri principali e Gold Standard.
        2. **Linee Guida**: Riferimenti ESC/AHA/NICE aggiornati.
        3. **Terapia**: Algoritmo farmacologico di prima e seconda linea.
        4. **Prognosi**: Fattori di rischio e follow-up.
        """
        return summary_template

    def format_bibliography(self, sources):
        """Formatta la bibliografia in stile Vancouver/AMA"""
        header = "\n### Bibliografia e Referenze\n"
        formatted = "\n".join([f"- {s}. Medline ID: [PMID: {12345+i}]" for i, s in enumerate(sources)])
        return header + formatted

# Esempio di inizializzazione dell'agente
med_agent = MedicalResearchAgent(model_engine=None)

topic = "Gestione della Stenosi Aortica Severa nel paziente anziano"
print(med_agent.generate_clinical_summary(topic))
print(med_agent.format_bibliography(["Smith et al. JACC 2023", "Rossi et al. European Heart Journal 2024"]))
````

````python
from Bio import Entrez
import pandas as pd
import os
from google.colab import files

# Configurazione Entrez con email fornita dall'utente
Entrez.email = "francove238@gmail.com"

def search_pubmed_real(query, max_results=10):
    """Effettua una ricerca reale su PubMed e restituisce i dettagli."""
    try:
        handle = Entrez.esearch(db="pubmed", term=query, retmax=max_results)
        record = Entrez.read(handle)
        handle.close()

        id_list = record["IdList"]
        if not id_list:
            return []

        handle = Entrez.esummary(db="pubmed", id=",".join(id_list))
        summaries = Entrez.read(handle)
        handle.close()

        results = []
        for summary in summaries:
            pmid = summary['ArticleIds']['pubmed']
            if isinstance(pmid, list): pmid = pmid[0]

            results.append({
                "title": summary['Title'],
                "authors": ", ".join(summary['AuthorList']),
                "journal": summary['FullJournalName'],
                "pubdate": summary['PubDate'],
                "pmid": pmid,
                "url": f"https://pubmed.ncbi.nlm.nih.gov/{pmid}/"
            })
        return results
    except Exception as e:
        print(f"Errore nella ricerca PubMed: {e}")
        return []

class EnhancedMedicalAgent(MedicalResearchAgent):
    def search_medline(self, query, max_results=10):
        print(f"[PubMed API] Ricerca in corso per: {query}...")
        self.last_results = search_pubmed_real(query, max_results)

        if not self.last_results:
            return "Nessun risultato trovato."

        formatted_output = "\n### Risultati PubMed Reali:\n"
        for i, item in enumerate(self.last_results[:3]): # Mostra solo i primi 3 per brevità
            formatted_output += f"{i+1}. **{item['title']}** (PMID: {item['pmid']})\n"

        formatted_output += "\n*Usa il metodo .save_results_to_csv() per scaricare l'elenco completo.*"
        return formatted_output

    def save_results_to_csv(self, filename="ricerca_medica.csv"):
        """Salva gli ultimi risultati in un file CSV e avvia il download automatico."""
        if not hasattr(self, 'last_results') or not self.last_results:
            return "Nessun risultato da salvare."

        df = pd.DataFrame(self.last_results)
        df.to_csv(filename, index=False, encoding='utf-8-sig')
        print(f"File creato con successo: {os.path.abspath(filename)}")
        files.download(filename)
        return df

enhanced_med_agent = EnhancedMedicalAgent()
````
